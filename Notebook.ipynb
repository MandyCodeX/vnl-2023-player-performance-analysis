```python
numeric_col = df.select_dtypes(include=['int','float']).columns
corr_matrix = df[numeric_col].corr()
corr_matrix
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix,annot=True,cmap='coolwarm',linewidth=5)
plt.title('correlation matrix heatmap')
plt.show()
```

```python
position_counts = df['Position'].value_counts()
position_counts.index
```

```python
explode= [0.1,0,0,0,0]
plt.pie(position_counts,explode=explode,labels=position_counts.index,autopct='%1.1f%%',startangle= 120,wedgeprops={'edgecolor': 'black', 'linewidth': 1, 'linestyle': '--'})
plt.title("Distribution of Position")
plt.show()
```

```python
avg_attack_by_country=df.groupby('Country')['Attack'].mean()
avg_attack_by_country.sort_values(ascending=False)
```

```python
avg_attack_by_country.sort_values(ascending=False).head(6).plot(kind='bar',color='#27C8F5')
plt.title("Average Attack By Country")
plt.xlabel('country')
plt.ylabel('Avg Attack')
```

```python
avg_serve_by_age = df.groupby('Age')['Serve'].mean()
avg_serve_by_age.sort_values(ascending=False)
```

```python
plt.hist(df['Age'],bins=20,color='skyblue',edgecolor='black')
plt.title("Distribution of Age")
plt.xlabel('Age')
plt.ylabel("Frequency")
plt.show()
```

```python
avg_attack_by_position=df.groupby('Position')['Attack'].mean()
avg_attack_by_position.sort_values(ascending=True).plot(kind='bar',color='m')
plt.title("Average Attack by Position")
plt.xlabel("Position")
plt.ylabel("Average Attack")
plt.show()
```

```python
serve_trend_by_age = df.groupby("Age")["Serve"].mean()
serve_trend_by_age.plot(kind='line',marker='o',linestyle='--',color='r')
plt.title("Serve Trend Over Age Group")
plt.xlabel("Age")
plt.ylabel("Average Serve")
plt.show()
```

```python
total_attack_block_by_country = df.groupby('Country')[['Attack','Block']].sum()
total_attack_block_by_country
```

```python
total_attack_block_by_country.sort_values(ascending=False,by="Attack").plot(kind='bar',stacked=True,cmap='cividis')
plt.title("Total Attack and Block by Country")
plt.xlabel("Country")
plt.ylabel("Total Value")
plt.show()
```
